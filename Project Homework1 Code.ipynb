## Binary Classification

```python
# Creating a binary classification target variable, whether the youth has ever used cigarettes or not 

binary_data = data.dropna(subset=['CIG_USED']) 
print(binary_data['CIG_USED'].value_counts())
print("\nTarget variable: CIG_USED\n(1: Used cigarettes, 0: Never used cigarettes)\n")
```

```python
# Splitting the data into training and testing sets
from sklearn.model_selection import train_test_split

X = binary_data.drop(columns=['CIG_USED', 'IRCIGFM'])  # Drop target and raw frequency
y = binary_data['CIG_USED']

# One-hot encode categorical feature (optional)
X = pd.get_dummies(X, drop_first=True)
```

```python
# Train-test split (80/20)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
print(f"\nTraining set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
print(f"Target distribution in train:\n{y_train.value_counts(normalize=True)}")
```

```python
# Visual representation of the data

# Count occurrences of each class in 'CIG_USED'
class_counts = binary_data['CIG_USED'].value_counts().sort_index()  # ensure 0 comes before 1

# Plot
plt.figure(figsize=(6, 4))
plt.bar([0, 1], class_counts, color=['#66c2a5', '#fc8d62'])  # similar to 'Set2' palette
plt.xticks([0, 1], ['Non-User (0)', 'User (1)'])
plt.title("Class Distribution: CIG_USED")
plt.ylabel("Number of Respondents")
plt.xlabel("Cigarette Use Status")
plt.tight_layout()
plt.show()
```

```python
# Training the decision tree 

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay
```

```python
clf = DecisionTreeClassifier(random_state=42, max_depth=5)             # Max depth can be adjusted
clf.fit(X_train, y_train)                                              # Train the model
y_pred = clf.predict(X_test)
```

```python
# Confusion matrix
ConfusionMatrixDisplay.from_estimator(clf, X_test, y_test, display_labels=["Non-User", "User"], cmap='Blues')
plt.title("Confusion Matrix")
plt.show()
```

```python
# Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=["Non-User", "User"]))
```

```python
# Decision tree visualization
from sklearn.tree import plot_tree

plt.figure(figsize=(10,5))
plot_tree(clf,
          feature_names=X_train.columns,
          class_names=["Non-User", "User"],
          filled=True,
          rounded=True,
          fontsize=8)
plt.title("Decision Tree for Cigarette Use (Binary Classification)")
plt.show()
```

```python
# ROC (Receiver Operating Characteristic) curve 
from sklearn.metrics import roc_curve, auc
```

```python
# Predicting probabilities for the positive class (if user = 1)
y_proba = clf.predict_proba(X_test)[:, 1]
```

```python
# Values
fpr, tpr, thresholds = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)
```

```python
# Plot
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f"ROC Curve (AUC = {roc_auc:.2f})", color='darkorange')
plt.plot([0, 1], [0, 1], 'k--', label="Random Classifier")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Binary Classification")
plt.legend(loc="lower right")
plt.grid(True)
plt.show()

'''Used this plot to check for the trade-off between true positive and false positive rate at different thresholds.
This helps in understanding the model's performance and selecting an optimal threshold for classification.'''
```

## Multi-Class Classification

```python
# Cleaning and preparing the data for multi-class classification
multi_data = data.copy()
```

```python
multi_data = multi_data[~multi_data['IRMJFM'].isin([94, 97, 98, 99])]
```

```python
# Convert to categorical for classification
multi_data['marij_use'] = multi_data['IRMJFM'].astype('category')
```

```python
# Check classes
print("Unique values in 'marij_use' (multi-class target):\n", sorted(multi_data['marij_use'].unique()))
```

```python
# Frequency distribution
print("\nMarijuana Use Frequency Distribution (before grouping):")
print(multi_data['marij_use'].value_counts())
```

```python
# Grouping the data for better interpretation 
def categorize_mj_use(x):
    if x <= 5:
        return "Low"
    elif 6 <= x <= 15:
        return "Moderate"
    elif 16 <= x <= 25:
        return "Frequent"
    elif 25 < x < 90 :
        return "Very Frequent"
    elif x == 91:
        return "NEVER USED MARIJUANA"
    elif x == 93:
        return "DID NOT USE MARIJUANA PAST MONTH"
    else:
        return np.nan
```

```python
multi_data['marij_use'] = multi_data['IRMJFM'].apply(categorize_mj_use)
multi_data['marij_use'] = multi_data['marij_use'].astype('category')
```

```python
# Ranges for marijuana use frequency                        
print("\nRanges for marijuana use frequency:")                          # For better reference/understanding
print("Low: 0-5 days")
print("Moderate: 6-15 days")
print("Frequent: 16-25 days") 
print("Very Frequent: >25 days")

print("\nMarijuana Use Frequency Distribution (after grouping):")
print(multi_data['marij_use'].value_counts())
```

```python
# One-Hot encoding categorical features for further analysis

# Separating features from target
multi_data = multi_data.dropna(subset=['marij_use'])                    # Dropping rows with NaN in marij_use

X = multi_data.drop(columns=['marij_use'])
y = multi_data['marij_use']
```

```python
# Getting categorical features (like gender, race, etc.)
X_encoded = pd.get_dummies(X)
```

```python
# Train-Test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y)

print("\nTrain shape:", X_train.shape)
print("Test shape:", X_test.shape)
print("\nTarget distribution in train:\n", y_train.value_counts(normalize=True))
```

```python
# Training the Multi-Class Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

```python
clf_multi = DecisionTreeClassifier(max_depth=5, random_state=42)
clf_multi.fit(X_train, y_train)
y_pred_multi = clf_multi.predict(X_test)

print("\nClassification Report (Multi-Class):")
print(classification_report(y_test, y_pred_multi))
```

```python
# Confusion Matrix
disp = ConfusionMatrixDisplay.from_predictions(y_test, y_pred_multi, cmap="Blues", xticks_rotation=45)
plt.title("Confusion Matrix - Marijuana Use Frequency")
plt.tight_layout()
plt.show()
```

```python
# Visualising the Decision Tree
from sklearn.tree import plot_tree

plt.figure(figsize=(10, 10))
plot_tree(clf_multi, 
          feature_names=X_train.columns, 
          class_names=clf_multi.classes_, 
          filled=True, 
          rounded=True, 
          fontsize=10)
plt.title("Decision Tree - Marijuana Use Frequency")
plt.tight_layout()
plt.show()
```

## Regression

```python
# Libraries
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# Cleaning and preparing the data for regression
regression_data = data.dropna(subset=['ALCYDAYS']).copy()               # Dropping rows with NaN in ALCYDAYS

X = regression_data.drop(columns=['ALCYDAYS'])
y = regression_data['ALCYDAYS']

X_encoded = pd.get_dummies(X, drop_first=True)
```

```python
# Train-Test data  split
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

print("\nTrain shape:", X_train.shape)
print("Test shape:", X_test.shape)
print("\nTarget distribution in train:\n", y_train.describe())
```

```python
# Regression Models
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
# Decision Tree Regressor
reg_tree = DecisionTreeRegressor(max_depth=5, random_state=42)
reg_tree.fit(X_train, y_train)
y_pred_tree = reg_tree.predict(X_test)
```

```python
# Evaluation metrics
mse_tree = mean_squared_error(y_test, y_pred_tree)
r2_tree = r2_score(y_test, y_pred_tree)
print("\nDecision Tree Regression Results:")
print(f"MSE: {mse_tree:.4f}")
print(f"R^2 Score: {r2_tree:.4f}")
```

```python
# Plotting true vs predicted values for Decision Tree
plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred_tree, alpha=0.5, color='teal')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
plt.xlabel("Actual ALCYDAYS")
plt.ylabel("Predicted ALCYDAYS")
plt.title("Decision Tree Regression: Actual vs Predicted")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from sklearn.impute import SimpleImputer

# Impute missing values with mean
imputer = SimpleImputer(strategy='mean')
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
# Linear Regression
lin_reg = LinearRegression()
lin_reg.fit(X_train_imputed, y_train)
y_pred_lin = lin_reg.predict(X_test_imputed)
```

```python
# Evaluation metrics for Linear Regression
mse_lin = mean_squared_error(y_test, y_pred_lin)
r2_lin = r2_score(y_test, y_pred_lin)
print("\nLinear Regression Results:")
print(f"MSE: {mse_lin:.4f}")
print(f"R^2 Score: {r2_lin:.4f}")
```

```python
# Plotting true vs predicted values for Linear Regression
plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred_lin, alpha=0.5, color='orange')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'b--')
plt.xlabel("Actual ALCYDAYS")
plt.ylabel("Predicted ALCYDAYS")
plt.title("Linear Regression: Actual vs Predicted")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Random Forest Regression (OPTIONAL!!)
from sklearn.ensemble import RandomForestRegressor

rf_reg = RandomForestRegressor(n_estimators=100, max_depth=6, random_state=42)
rf_reg.fit(X_train_imputed, y_train)
y_pred_rf = rf_reg.predict(X_test_imputed)
```

```python
# Evaluation metrics
mse_rf = mean_squared_error(y_test, y_pred_rf)
r2_rf = r2_score(y_test, y_pred_rf)
print("\nRandom Forest Regression Results:")
print(f"MSE: {mse_rf:.4f}")
print(f"R^2 Score: {r2_rf:.4f}")
```

```python
# Plotting true vs predicted values
plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred_rf, alpha=0.5, color='purple')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
plt.xlabel("Actual ALCYDAYS")
plt.ylabel("Predicted ALCYDAYS")
plt.title("Random Forest Regression: Actual vs Predicted")
plt.grid(True)
plt.tight_layout()
plt.show()
```
